## Inputs outside the Task 2 `scan_db` design limits

```python
limits = {"r_over_a": (0.25, 0.75), "RLTS_1": (1.5, 3.5), "RLTS_2": (1.5, 3.5), "RLNS_1": (1, 3),
          "TAUS_2": (0.5, 1.5), "XNUE": (0.01, 1)}
outside = pd.DataFrame({k: ~team[k].between(lo, hi) for k, (lo, hi) in limits.items()})
print(f"{outside.any(axis=1).sum():,} of {len(team):,} cases ({outside.any(axis=1).mean():.1%}) are outside at least one limit.")
outside.sum().rename("cases outside").to_frame()
```
