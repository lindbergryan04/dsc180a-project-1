# Task 7 results: transfer to DIII-D shot 193870

Reads `results/transfer_scores.csv`. The models were trained on shot 128913 only and were frozen before this evaluation. Scores compare with numerical TGLF for each cohort.

```python
from pathlib import Path
import pandas as pd

pd.set_option("display.precision", 3)
HERE = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "notebooks/ryan").is_dir()) / "notebooks/ryan/task7"
results = HERE / "results"
```

## Full-data original-width models: R² by cohort

```python
scores = pd.read_csv(results / "transfer_scores.csv")
full = scores.loc[scores.family.eq("Task 6") & scores.train_size.eq(16586) & scores.width.eq("original")]
full.pivot(index="target", columns="cohort", values="r2")
```

## All 81 Task 6 ensembles on shot 193870

Range of R² across training sizes, draws, and widths.

```python
new = scores.loc[scores.family.eq("Task 6") & scores.cohort.str.startswith("193870")]
new.groupby(["target", "cohort"]).r2.agg(["min", "median", "max", "count"])
```

Mean R² on the 517 perturbed shot-193870 cases by training size (original width).
